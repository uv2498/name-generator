## 1. Packages

If a line says `MISSING` or `PROBLEM`, add a new cell below this one, paste the command shown next to it, run it, and then choose **Restart** and **Run All**.

```python
import importlib
import sys
import warnings

warnings.filterwarnings("ignore")                 # hide technical messages

print(f"Python {sys.version_info.major}.{sys.version_info.minor}")
problems = []
for package in ["torch", "matplotlib", "transformers"]:
    try:
        module = importlib.import_module(package)
        print(f"  OK       {package} {module.__version__}")
    except Exception:
        print(f"  MISSING  {package}   ->  run in a new cell:  %pip install {package}")
        problems.append(package)

if "torch" not in problems:
    try:
        import torch
        torch.ones(2).numpy()                     # PyTorch and NumPy must work together
        print("  OK       torch works together with numpy")
    except Exception:
        print("  PROBLEM  torch cannot use numpy   ->  run in a new cell:  %pip install --upgrade numpy torch")
        problems.append("numpy")

if sys.version_info < (3, 10):
    print("  PROBLEM  Python 3.10 or newer is needed. Install a newer Python and select it in VS Code.")
    problems.append("python")
```

## 2. Workshop files

This notebook must sit in the workshop folder, next to the folders `data` and `app`. In VS Code, open the **whole folder** (File > Open Folder), not a single file.

## 3. Download the model for Part 3

This downloads a small language model from Hugging Face (about 270 MB) and asks it one question. The download happens only once; the model is then stored on your computer.

```python
MODEL = "HuggingFaceTB/SmolLM2-135M-Instruct"

if "transformers" in problems or "torch" in problems:
    print("Install the missing packages first (Step 1), then run this cell again.")
else:
    try:
        import transformers
        from transformers import pipeline

        transformers.logging.set_verbosity_error()
        print("Downloading and loading the model. This can take a few minutes the first time ...")
        chat = pipeline("text-generation", model=MODEL)
        result = chat([{"role": "user", "content": "Say hello in five words."}], max_new_tokens=20)
        print("  OK       the model answered:", result[0]["generated_text"][-1]["content"].strip())
    except Exception as error:
        print("  PROBLEM  the model could not be loaded:", type(error).__name__, "-", str(error)[:200])
        print("           1. Check your internet connection and run this cell again.")
        print("           2. If it still fails, run in a new cell:  %pip install --upgrade transformers")
        print("              then choose Restart and Run All.")
        problems.append("model")
```
